# Introduction to Digital Signal Processing

Digital Signal Processing (DSP) is a branch of engineering that focuses on the manipulation and analysis of digital signals using algorithms and computational techniques. These signals can be of various types, such as audio, video, images, or biomedical data, and are generally represented as functions of one or more independent variables. The goal of DSP is to enhance, modify, or extract useful information from these signals.

## Mathematical Foundations of DSP

The **Hilbert space** formalism underpins the representation and analysis of signals by providing a unified geometric framework where signals are treated as **vectors**. This perspective allows the extension of the geometric intuition of traditional Euclidean spaces to spaces of *abstract signals*, whether discrete sequences or continuous functions.

The pillars of this formalism are detailed below:

1. **Signals as vectors in metric spaces**
In this framework, finite-energy signals are represented in specific spaces: $L_2(\mathbb{R})$ for continuous-time functions and $\ell_2(\mathbb{Z})$ for discrete-time sequences.

The concept of a norm assigns a "length" or size to the signal, which is equivalent to its physical energy.

The completeness of Hilbert space guarantees that any sequence of signals that appears to converge (Cauchy sequences) has a limit that is also a valid signal within the same space, which is critical for the consistency of the mathematical analysis.

2. **The inner product and the notion of similarity**
The inner product is the fundamental tool for measuring the relationship between two signals.

- **Similarity**: It functions as a measure of how similar two signals are; a large inner product indicates a similar orientation or shape, while a value of zero indicates that the signals are orthogonal (uncorrelated).
- **Information Extraction**: Allows obtaining representation coefficients through the inner product between the original signal and vectors of a basis

3. **Representations using bases and frames**
The formalism allows decomposing a complex signal into a sum of simpler components using a coordinate system or basis

- **Orthonormal Bases**: These are ideal because they simplify the calculation of coefficients and preserve the signal's energy between the time domain and the transformed domain, a principle known as Parseval's equality
- **Fourier Representations**: Fourier analysis is based on the fact that complex exponentials are eigenfunctions of linear time-invariant (LSI) systems
This means that the Hilbert space allows "diagonalizing" convolution operations, converting complex processes into simple multiplications in the frequency domain

4. **Optimal Approximation and the Projection Theorem**
One of the most powerful applications in signal analysis is optimal approximation.

The Projection Theorem states that the best approximation of a signal within a subspace (e.g., a band-limited or compressed version) is its orthogonal projection onto that subspace. This minimizes the mean squared error, underpinning essential techniques of filtering, noise reduction, and compression.


## Hilbert Spaces at a First Glance

Mathematically, a Hilbert space is defined as an inner product space that is complete with respect to the metric induced by that product. For a set of elements (vectors) to constitute a Hilbert space, it must satisfy the following fundamental conditions:

1. **Structure of an Inner Product Space**
The space must be equipped with a function called the inner product $\langle \cdot,\cdot \rangle$, which assigns a scalar to each pair of vectors and satisfies the following properties:

- **Linearity**: It is linear in its first argument ($\langle \alpha x, y \rangle = \alpha\langle x, y \rangle$)
- **Hermitian Symmetry**: The product of two elements is equal to the complex conjugate of its inverse ($\langle x,y \rangle = \langle y,x \rangle^*$)
- **Definite Positivity**: The product of a vector with itself is always non-negative ($\langle x,x \rangle \geq 0$) and is only zero if the vector is the zero vector
- **Induced Norm**: This inner product automatically defines a norm or "length" of the vector through the relation $\|x\| = \sqrt{\langle x,x \rangle}$

2. **Completeness (Critical Condition)**
The characteristic that distinguishes a Hilbert space from a simple inner product space is completeness.

This means that the space is a complete metric space, which implies:

- **Cauchy sequences**: Any sequence of vectors where the elements approach each other arbitrarily closely (Cauchy sequence) must converge to a limit.
- **Limit within the space**: The value of the limit of such a sequence must be a vector that also belongs to the same space.
- **If an inner product space is not complete, it is simply called a pre-Hilbert space.**
    
3. **Geometric Properties and Implications**
Thanks to these conditions, Hilbert spaces allow us to extend Euclidean geometric intuition (such as that of $\mathbb{R}^2$ or $\mathbb{R}^3$) to infinite-dimensional abstract spaces, such as those of functions or sequences.

This guarantees fundamental properties such as:

- **Orthogonality**: Two vectors are orthogonal if their inner product is zero.
- **Projection Theorem**: In a Hilbert space, there always exists a unique element in a closed subspace that is the closest to a given vector outside that subspace.
- **Metric Relations**: The parallelogram identity and the Pythagorean theorem hold.

In summary, a Hilbert space provides the necessary mathematical framework for consistent signal analysis (treating signals as vectors), ensuring that operations such as Fourier series expansions or least-squares approximations always have a valid solution within the space.
